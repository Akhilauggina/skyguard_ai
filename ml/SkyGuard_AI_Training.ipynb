```python
def parse_temperature(value):
    if pd.isna(value):
        return np.nan

    temp = value.split(",")[0]

    if temp == "+9999":
        return np.nan

    return int(temp) / 10
```

```python
df["temperature"] = df["TMP"].apply(parse_temperature)
```

```python
def parse_dew(value):
    if pd.isna(value):
        return np.nan

    dew = value.split(",")[0]

    if dew == "+9999":
        return np.nan

    return int(dew) / 10
```

```python
df["dew_point"] = df["DEW"].apply(parse_dew)
```

```python
def parse_pressure(value):
    if pd.isna(value):
        return np.nan

    pressure = value.split(",")[0]

    if pressure == "99999":
        return np.nan

    return int(pressure) / 10
```

```python
df["pressure"] = df["SLP"].apply(parse_pressure)
```

```python
def parse_wind(value):
    if pd.isna(value):
        return pd.Series([np.nan, np.nan])

    parts = value.split(",")

    direction = parts[0]
    speed = parts[3]

    if direction == "999":
        direction = np.nan
    else:
        direction = int(direction)

    if speed == "9999":
        speed = np.nan
    else:
        speed = int(speed) / 10

    return pd.Series([direction, speed])
```

```python
df[["wind_direction", "wind_speed"]] = df["WND"].apply(parse_wind)
```

```python
def parse_visibility(value):
    if pd.isna(value):
        return np.nan

    vis = value.split(",")[0]

    if vis == "999999":
        return np.nan

    return int(vis)
```

```python
df["visibility"] = df["VIS"].apply(parse_visibility)
```

```python
df[
    [
        "temperature",
        "dew_point",
        "pressure",
        "wind_speed",
        "wind_direction",
        "visibility",
    ]
].head()
```

```python
def calculate_relative_humidity(temp, dew):
    if pd.isna(temp) or pd.isna(dew):
        return np.nan

    a = 17.625
    b = 243.04

    rh = 100 * (
        np.exp((a * dew) / (b + dew)) /
        np.exp((a * temp) / (b + temp))
    )

    return round(rh, 2)
```

```python
df["humidity"] = df.apply(
    lambda row: calculate_relative_humidity(
        row["temperature"],
        row["dew_point"]
    ),
    axis=1
)
```

### Reconstructing `df_model` with all available pressure data

```python
import pandas as pd

# Define df_full by concatenating the original dataframes, as df_full was not defined here.
df_full = pd.concat([df23, df24, df25], ignore_index=True)

# Re-apply all parsing functions to df_full
df_processed = df_full.copy()

df_processed["temperature"] = df_processed["TMP"].apply(parse_temperature)
df_processed["dew_point"] = df_processed["DEW"].apply(parse_dew)
df_processed["pressure"] = df_processed["SLP"].apply(parse_pressure)
df_processed[["wind_direction", "wind_speed"]] = df_processed["WND"].apply(parse_wind)
df_processed["visibility"] = df_processed["VIS"].apply(parse_visibility)

# Calculate humidity
df_processed["humidity"] = df_processed.apply(
    lambda row:
        calculate_relative_humidity(
            row["temperature"],
            row["dew_point"]
        ),
    axis=1
)

# Convert DATE and extract time components
df_processed["DATE"] = pd.to_datetime(df_processed["DATE"])
df_processed["year"] = df_processed["DATE"].dt.year
df_processed["month"] = df_processed["DATE"].dt.month
df_processed["day"] = df_processed["DATE"].dt.day
df_processed["hour"] = df_processed["DATE"].dt.hour
df_processed["day_of_week"] = df_processed["DATE"].dt.dayofweek

print("Features re-processed for df_full.")
```

```python
# Define the features for the final df_model
features = [
    "temperature",
    "humidity",
    "pressure",
    "wind_speed",
    "wind_direction",
    "visibility",
    "month",
    "hour"
]

# Create the new df_model with all available data
df_model_updated = df_processed[features].copy()

print("Updated df_model created.")
display(df_model_updated.head())
print("\nNaN counts in the updated df_model:")
display(df_model_updated.isnull().sum())
```

### Handling Missing Values in `pressure` using Interpolation

```python
# Apply linear interpolation to the 'pressure' column
df_model_updated['pressure'] = df_model_updated['pressure'].interpolate(method='linear')

print("NaN counts in df_model_updated after pressure interpolation:")
display(df_model_updated.isnull().sum())
```

### Checking Pressure Values in Other Report Types

```python
# Re-concatenate the original dataframes to restore all columns
df_full = pd.concat([df23, df24, df25], ignore_index=True)

# Filter for REPORT_TYPEs other than 'FM-15' from the full DataFrame
df_other_reports = df_full[df_full['REPORT_TYPE'] != 'FM-15'].copy()

print(f"Shape of DataFrame with other report types: {df_other_reports.shape}")
print(f"Unique REPORT_TYPEs in this subset: {df_other_reports['REPORT_TYPE'].unique()}")
```
